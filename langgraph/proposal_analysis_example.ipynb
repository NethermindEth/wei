## Example Proposal

Let's define an example proposal to analyze.

```python
# Example Proposal
example_proposal = """
# EIP-1234: Ethereum Network Upgrade - Proof of Stake Transition

## Abstract
This proposal outlines a comprehensive plan for transitioning the Ethereum network from Proof of Work (PoW) to Proof of Stake (PoS) consensus mechanism. The transition aims to significantly reduce energy consumption, improve scalability, and enhance security of the network.

## Motivation
The current PoW consensus mechanism has several limitations:
1. High energy consumption and environmental impact
2. Limited scalability with increasing transaction volumes
3. Potential centralization due to mining hardware requirements

By transitioning to PoS, Ethereum can address these issues while maintaining decentralization and security.

## Specification
The transition will occur in three phases:

### Phase 1: Beacon Chain Deployment
- Launch a parallel PoS chain (Beacon Chain)
- Allow ETH holders to become validators by staking their ETH
- Minimum stake requirement: 32 ETH

### Phase 2: The Merge
- Merge the existing PoW chain with the Beacon Chain
- Transition block production to PoS validators
- Maintain backward compatibility for existing applications

### Phase 3: Sharding Implementation
- Implement sharding to improve scalability
- Split the network into 64 shard chains
- Enable parallel transaction processing

## Security Considerations
- Slashing mechanisms to penalize malicious validators
- Minimum stake requirements to prevent Sybil attacks
- Decentralization metrics will be continuously monitored

## Backward Compatibility
All existing applications will continue to function without modification after the transition.
"""

example_metadata = {
    "id": "EIP-1234",
    "title": "Ethereum Network Upgrade - Proof of Stake Transition",
    "author": "Vitalik Buterin",
    "date_submitted": "2022-03-15",
    "protocol": "Ethereum",
    "category": "Core",
    "url": "https://eips.ethereum.org/EIPS/eip-1234"
}
```

## Run the Analysis

Now let's run the proposal analysis and display the results.

```python
# Run the analysis
print("Starting proposal analysis...")
print(f"Using Exa API for search and fetching")
print(f"Using OpenRouter API with gpt-4o-mini for analysis components")
print(f"Using OpenRouter API with perplexity/sonar-pro for roadmap modeling")

analysis_result = analyze_proposal(example_proposal, example_metadata)
```

## Display Results

Let's display the results in a structured format.

```python
from IPython.display import display, Markdown, HTML

# Display the results using HTML and Markdown for better formatting
display(Markdown("# Proposal Analysis Results"))
```

```python
# Display tasks
display(Markdown("## Tasks"))
for i, task in enumerate(analysis_result["tasks"]):
    task_html = f"""
    <div style="margin-bottom: 15px; border-left: 3px solid #007bff; padding-left: 10px;">
        <h4>Task {i+1}: {task['description']}</h4>
        <p><strong>Priority:</strong> {task['priority']}</p>
    """
    
    if 'evidence' in task and task['evidence']:
        task_html += "<p><strong>Evidence:</strong></p><ul>"
        for evidence in task['evidence']:
            task_html += f"<li>{evidence}</li>"
        task_html += "</ul>"
        
    task_html += "</div>"
    display(HTML(task_html))
```

```python
# Display blockers
display(Markdown("## Blockers"))
blockers_html = "<ul>"
for blocker in analysis_result["blockers"]:
    blockers_html += f"<li>{blocker}</li>"
blockers_html += "</ul>"
display(HTML(blockers_html))
```

```python
# Display evidence summary
display(Markdown("## Evidence Summary"))
display(Markdown(analysis_result["evidence_summary"]))
```

```python
# Display next steps
display(Markdown("## Next Steps"))
steps_html = "<ol>"
for step in analysis_result["next_steps"]:
    steps_html += f"<li>{step}</li>"
steps_html += "</ol>"
display(HTML(steps_html))
```

```python
# Display comprehensive evaluation report
if "evaluation_report" in analysis_result and analysis_result["evaluation_report"]:
    display(Markdown("# Comprehensive Evaluation Report"))
    report = analysis_result["evaluation_report"]
    
    display(Markdown(f"**Summary:** {report['summary']}"))
    
    display(Markdown("## Evaluation Criteria"))
    
    criteria = [
        ("Goals & Motivation", report["goals_and_motivation"]),
        ("Measurable Outcomes", report["measurable_outcomes"]),
        ("Budget", report["budget"]),
        ("Technical Specifications", report["technical_specifications"]),
        ("Language Quality", report["language_quality"])
    ]
    
    for name, criterion in criteria:
        status = criterion["status"].upper()
        status_icon = "✅" if status == "PASS" else "❌" if status == "FAIL" else "⚠️"
        
        criterion_html = f"""
        <div style="margin-bottom: 20px; padding: 10px; border-radius: 5px;">
            <h3>{name}: {status_icon} {status}</h3>
        """
        
        if criterion["justification"]:
            criterion_html += f"<p><strong>Justification:</strong> {criterion['justification']}</p>"
            
        if criterion["suggestions"]:
            criterion_html += "<p><strong>Suggestions:</strong></p><ul>"
            for suggestion in criterion["suggestions"]:
                criterion_html += f"<li>{suggestion}</li>"
            criterion_html += "</ul>"
            
        criterion_html += "</div>"
        display(HTML(criterion_html))
```

```python
# Display arguments for and against
display(Markdown("## Arguments"))

arguments_html = """
<div style="display: flex; justify-content: space-between;">
    <div style="width: 48%; padding: 10px; border-radius: 5px;">
        <h3>For the Proposal</h3>
        <ul>
"""

for arg in report["arguments"]["for_proposal"]:
    arguments_html += f"<li>✓ {arg}</li>"
    
arguments_html += """
        </ul>
    </div>
    <div style="width: 48%; padding: 10px; border-radius: 5px;">
        <h3>Against the Proposal</h3>
        <ul>
"""

for arg in report["arguments"]["against"]:
    arguments_html += f"<li>✗ {arg}</li>"
    
arguments_html += """
        </ul>
    </div>
</div>
"""

display(HTML(arguments_html))
```

## Claim-Evidence Graph Visualization

Let's visualize the claim-evidence graph to better understand the relationships between claims and their supporting evidence.

```python
import networkx as nx
import matplotlib.pyplot as plt

# Create a sample claim-evidence graph based on the analysis results
G = nx.DiGraph()

# Extract claims from the analysis result
if "claims_evidence" in analysis_result:
    claims_evidence = analysis_result["claims_evidence"]
    
    # Add claim nodes
    for i, claim_evidence in enumerate(claims_evidence):
        claim_id = f"claim{i+1}"
        G.add_node(claim_id, type="claim", content=claim_evidence["claim"], confidence=claim_evidence.get("confidence", 0.5))
        
        # Add evidence nodes and edges
        for j, evidence in enumerate(claim_evidence.get("evidence", [])):
            evidence_id = f"evidence{i+1}_{j+1}"
            source = claim_evidence.get("sources", ["unknown"])[j] if j < len(claim_evidence.get("sources", [])) else "unknown"
            G.add_node(evidence_id, type="evidence", content=evidence, source=source)
            G.add_edge(evidence_id, claim_id, type="supports")
else:
    # Create a sample graph if no claims_evidence is available
    G.add_node("claim1", type="claim", content="PoS reduces energy consumption", confidence=0.9)
    G.add_node("claim2", type="claim", content="PoS improves scalability", confidence=0.8)
    G.add_node("claim3", type="claim", content="PoS maintains security", confidence=0.7)

    G.add_node("evidence1", type="evidence", content="Academic research shows 99% energy reduction", source="research_paper")
    G.add_node("evidence2", type="evidence", content="Testnet performance metrics", source="ethereum_blog")
    G.add_node("evidence3", type="evidence", content="Slashing mechanism design", source="eip_doc")
    G.add_node("evidence4", type="evidence", content="Parallel transaction processing capability", source="whitepaper")
    G.add_node("evidence5", type="evidence", content="Validator economic incentives", source="economic_analysis")

    G.add_edge("evidence1", "claim1", type="supports")
    G.add_edge("evidence2", "claim1", type="supports")
    G.add_edge("evidence2", "claim2", type="supports")
    G.add_edge("evidence4", "claim2", type="supports")
    G.add_edge("evidence3", "claim3", type="supports")
    G.add_edge("evidence5", "claim3", type="supports")

# Visualize the graph
plt.figure(figsize=(12, 8), facecolor='white')

# Set white background
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

# Define node colors - using very light colors with black borders for better readability
node_colors = ['#E6F3FF' if G.nodes[n]['type'] == 'claim' else '#F0FFF0' for n in G.nodes()]

# Create a layout for the graph
pos = nx.spring_layout(G, seed=42, k=0.3)  # Increase k for more spacing

# Draw edges
nx.draw_networkx_edges(G, pos, arrows=True, width=1.5, edge_color='gray')

# Draw nodes with white background and black borders
nx.draw_networkx_nodes(G, pos, node_color=node_colors, node_size=2000, 
                      edgecolors='black', linewidths=1.5)

# Add labels with black text
labels = {}
for node in G.nodes():
    if G.nodes[node]['type'] == 'claim':
        # Use full content for better understanding
        content = G.nodes[node]['content']
        if len(content) > 20:
            content = content[:17] + "..."
        labels[node] = f"{content}\n(conf: {G.nodes[node]['confidence']})"
    else:
        content = G.nodes[node]['content']
        if len(content) > 20:
            content = content[:17] + "..."
        labels[node] = f"{content}\n(src: {G.nodes[node]['source']})"

nx.draw_networkx_labels(G, pos, labels=labels, font_size=9, font_color='black', font_weight='bold')

# Add a legend with black text
from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], marker='o', color='w', markerfacecolor='#E6F3FF', markersize=15, label='Claims', markeredgecolor='black'),
    Line2D([0], [0], marker='o', color='w', markerfacecolor='#F0FFF0', markersize=15, label='Evidence', markeredgecolor='black')
]
plt.legend(handles=legend_elements, loc="upper right", frameon=True, facecolor='white', edgecolor='black')

# Add title with black text
plt.title("Claim-Evidence Graph", fontsize=14, fontweight='bold', color='black')

# Remove axis
plt.axis('off')

# Show the plot
plt.tight_layout()
plt.show()
```
